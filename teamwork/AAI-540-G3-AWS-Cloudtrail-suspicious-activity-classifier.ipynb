## 1.2 Download Dataset

```python
from pathlib import Path
import tarfile
import gzip 
import json
from datetime import datetime
import numpy as np 
import pandas as pd
import requests

# Create directories in current project
project_dir = Path.cwd()
raw_dir = project_dir / "raw"
processed_dir = project_dir / "processed"
model_dir = project_dir / "model"
report_dir = project_dir / "reports"

for path in [ raw_dir, processed_dir, model_dir, report_dir ]:
    path.mkdir(parents=True, exist_ok=True)

# Download
tar_file_path = raw_dir / "flaws_cloudtrail_logs.tar"
dataset_url = "https://summitroute.com/downloads/flaws_cloudtrail_logs.tar"
if not tar_file_path.exists():
    download_response = requests.get(dataset_url, stream=True)
    download_response.raise_for_status()
    with open(tar_file_path, "wb") as f:
        for chunk in download_response.iter_content(chunk_size=8192):
            f.write(chunk)
    print("Tar file downloaded:", tar_file_path)
else:
    print("Tar file already exists:", tar_file_path)

# Extract
with tarfile.open(tar_file_path, "r") as t:
    t.extractall(path=raw_dir)
print("Extraction completed. Files are in:", raw_dir)
```

## 1.4 Extract dataset

```python
gz_files_path = raw_dir / "flaws_cloudtrail_logs"
extracted_json_dir = raw_dir / "extracted"
extracted_json_dir.mkdir(parents=True, exist_ok=True)
for gz_file in gz_files_path.glob("*.json.gz"):
    json_file_path = extracted_json_dir / gz_file.with_suffix("").name
    if json_file_path.exists():
       continue

    with gzip.open(gz_file, "rt", encoding="utf-8") as gzf:
        json_content = json.load(gzf)
    with open(json_file_path, "w", encoding="utf-8") as json_file: 
        json.dump(json_content, json_file, indent=2)
    print(f"Decompressed and saved to {json_file_path}")
print("Extracted all json files.")
```

## 1.3 Parse CLoud Trail Records

```python
def get_nested(d, path, default=""):
    current = d
    for key in path:
        if not isinstance(current, dict):
            return default
        current = current.get(key, default)
    return current if current is not None else default


def flatten_cloudtrail_record(record: dict) -> dict:
    user_identity = record.get("userIdentity") or {}
    session_attrs = get_nested(user_identity, ["sessionContext", "attributes"], {})
    if not isinstance(session_attrs, dict):
        session_attrs = {}

    return {
        "event_id": record.get("eventID", ""),
        "event_time": record.get("eventTime", ""),
        "event_type": record.get("eventType", ""),
        "event_source": record.get("eventSource", ""),
        "event_name": record.get("eventName", ""),
        "aws_region": record.get("awsRegion", ""),
        "source_ip": record.get("sourceIPAddress", ""),
        "user_agent": record.get("userAgent", ""),
        "user_type": user_identity.get("type", ""),
        "principal_id": user_identity.get("principalId", ""),
        "account_id": user_identity.get("accountId", ""),
        "arn": user_identity.get("arn", ""),
        "access_key_id_present": int(bool(user_identity.get("accessKeyId"))),
        "mfa_authenticated": session_attrs.get("mfaAuthenticated", ""),
        "session_creation_date": session_attrs.get("creationDate", ""),
        "error_code": record.get("errorCode", ""),
        "error_message": record.get("errorMessage", ""),
        "request_parameters_present": int(record.get("requestParameters") is not None),
        "response_elements_present": int(record.get("responseElements") is not None),
    }


def iter_cloudtrail_records(gz_path: Path):
    with gzip.open(gz_path, "rt", encoding="utf-8") as f:
        data = json.load(f)
    for record in data.get("Records", []):
        yield flatten_cloudtrail_record(record)


def load_cloudtrail_files(files, max_records=None):
    rows = []
    for gz_path in files:
        for row in iter_cloudtrail_records(gz_path):
            rows.append(row)
            if max_records and len(rows) >= max_records:
                return pd.DataFrame(rows)
    return pd.DataFrame(rows)

# For quick development, start with a sample. Increase or remove max_records for final training.
gz_files_path= Path("/home/sagemaker-user/project/AWS_CloudTrail_Suspicious_Activity_Classifier/raw/flaws_cloudtrail_logs")
files = gz_files_path.glob("*.json.gz")
json_df = load_cloudtrail_files(files, max_records=250_000)
print(json_df.shape)
```

# 2. Exploratory Data Analysis

```python
df = json_df
df["event_time"] = pd.to_datetime(df["event_time"], errors="coerce", utc=True)

## Drop the events with missing values 
df = df.dropna(subset=["event_time"]).sort_values("event_time").reset_index(drop=True)

summary = {
    "rows": len(df),
    "start_time": df["event_time"].min(),
    "end_time": df["event_time"].max(),
    "unique_principals": df["principal_id"].nunique(),
    "unique_source_ips": df["source_ip"].nunique(),
    "unique_user_agents": df["user_agent"].nunique(),
    "unique_event_names": df["event_name"].nunique(),
}
summary
```

```python
# Top event sources 
display(df["event_source"].value_counts().head(15))
```

- The dataset is dominated by **EC2 (100,170 events)** and **STS (53,505 events)**, followed by S3 and IAM. This indicates that infrastructure activity, temporary credential operations, storage access, and identity management account for much of the recorded activity.
- CloudTrail contributes 4,496 events, making its API activity relevant to investigating logging configuration changes. However, the event names and context must be examined before interpreting these records as suspicious.
- The uneven service distribution may cause the model to learn common-service patterns more effectively than patterns from less frequent services.
- Evaluation should therefore examine performance by service where sufficient labeled examples exist.

```python
# Top event names
display(df["event_name"].value_counts().head(20))
```

- These counts describe the activity patterns.
- AssumeRole **(48,968 events) and DescribeSnapshots (40,595 events)** dominate the recorded activity. These counts indicate frequent role-assumption requests and snapshot queries, but do not show whether the requests succeeded or were authorized.
- Many other frequent events involve inspecting resources and permissions, including **DescribeInstances, ListBuckets, GetBucketAcl, and GetPolicyVersion**. Such actions can support routine administration or reconnaissance; frequency alone cannot distinguish the two.
- RunInstances (11,064 events) also represents substantial instance-launch activity that warrants examining identity, authorization errors, and timing.
- For classification, event names should be combined with identity type, MFA status, error codes, and other context.

```python
# User Identity Types
display(df["user_type"].value_counts().head(10))
```

- **IAMUser accounts for approximately 76% of events**, indicating that IAM-user identities dominate the dataset. AWSService and AssumedRole contribute about 11% and 9%, respectively.
- **Root accounts for 10,268 events (4.1%)**. Under this project’s labeling policy, all root activity receives a suspicious label. This is a review-priority rule, not evidence that every root event is malicious.
- AWSAccount activity is relatively uncommon, and **one record has a blank identity type**. Map that blank value to `unknown` during preprocessing.
- The uneven distribution means overall evaluation metrics may primarily reflect IAMUser activity. Review model performance by identity type where enough examples exist.

```python
# Error Codes
display(df["error_code"].replace("", "NO_ERROR").value_counts().head(15))
```

- NO_ERROR dominates with 189,162 events, indicating that most recorded requests have no reported error code. This does not necessarily mean the activity was safe.
- Authorization-related errors are prominent: AccessDenied (40,674) and Client.UnauthorizedOperation (6,045). These are useful risk signals, although legitimate permission issues can also produce them.
- The project’s labeling rules classify authorization failures as suspicious for review.
- Other errors need different interpretations:
  - InstanceLimitExceeded / InsufficientInstanceCapacity: resource limits or capacity constraints.
  - RequestLimitExceeded: request throttling.
  - NoSuchEntity / missing configuration errors: requested entities or configurations were not found.
  - DryRunOperation: a distinct dry-run outcome that should not be treated as an authorization failure.
    
For modeling, retain the specific error category and distinguish authorization failures from other errors rather than labeling every error as suspicious.

## 3. Labelling Strategy

Label definitions: 0 = Normal and 1 = suspicious

A record is labeled suspicious when it matches risky cloud behavior such as:

- root user activity without MFA
- CloudTrail logging disabled or deleted
- IAM access key or policy changes
- S3 bucket policy changes or bucket enumeration by high-risk identity
- repeated authorization failures
- activity from unusual source IP/region/user-agent behavior

In the design document, call this out clearly as **rule-generated weak labeling**. The ML model learns from these labels and produces a probability score.
